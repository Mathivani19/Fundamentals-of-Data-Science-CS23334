```python


import numpy as np
import pandas as pd

# Create the dataset directly in Python
data = {
    "CustomerID": [1, 2, 3, 4, 5, 6, 7, 8, 9, 9, 10],
    "Age_Group": ["20-25", "30-35", "25-30", "20-25", "35+",
                  "35+", "35+", "20-25", "25-30", "25-30", "30-35"],
    "Rating(1-5)": [4, 5, 6, -1, 3, 3, 4, 7, 2, 2, 5],
    "Hotel": ["Ibis", "LemonTree", "RedFox", "LemonTree", "Ibis",
              "Ibys", "RedFox", "LemonTree", "Ibis", "Ibis", "RedFox"],
    "FoodPreference": ["veg", "Non-Veg", "Veg", "Veg", "Vegetarian",
                       "Non-Veg", "Vegetarian", "Veg", "Non-Veg",
                       "Non-Veg", "non-Veg"],
    "Bill": [1300, 2000, 1322, 1234, 989, 1909, 1000, 2999, 3456, 3456, 6755],
    "NoOfPax": [2, 3, 2, 2, 2, 2, -1, -10, 3, 3, 4],
    "EstimatedSalary": [40000, 59000, 30000, 120000, 45000, 122220,
                        21122, 345673, -99999, -99999, 87777],
    "Age_Group.1": ["20-25", "30-35", "25-30", "20-25", "35+",
                    "35+", "35+", "20-25", "25-30", "25-30", "30-35"]
}

df = pd.DataFrame(data)

# Display original dataset
print("Original Dataset:")
print(df)

# Remove duplicate rows and reset index
df.drop_duplicates(inplace=True)
df.reset_index(drop=True, inplace=True)

# Remove extra column
df.drop(columns=["Age_Group.1"], inplace=True)

# Handle inappropriate values
df.loc[~df["Rating(1-5)"].between(1, 5), "Rating(1-5)"] = np.nan
df.loc[df["Bill"] < 0, "Bill"] = np.nan
df.loc[df["EstimatedSalary"] < 0, "EstimatedSalary"] = np.nan
df.loc[~df["NoOfPax"].between(1, 20), "NoOfPax"] = np.nan

# Correct hotel names
df["Hotel"] = df["Hotel"].replace({"Ibys": "Ibis"})

# Standardize food preferences
df["FoodPreference"] = df["FoodPreference"].replace({
    "veg": "Veg",
    "Vegetarian": "Veg",
    "non-Veg": "Non-Veg"
})

# Fill missing numerical values
df["Rating(1-5)"] = df["Rating(1-5)"].fillna(
    df["Rating(1-5)"].median()
)
df["Bill"] = df["Bill"].fillna(round(df["Bill"].mean()))
df["EstimatedSalary"] = df["EstimatedSalary"].fillna(
    round(df["EstimatedSalary"].mean())
)
df["NoOfPax"] = df["NoOfPax"].fillna(
    round(df["NoOfPax"].median())
)

# Display cleaned dataset
print("\nCleaned Dataset:")
print(df.to_string(index=False))

# Dataset information
print("\nDataset Information:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())
```
