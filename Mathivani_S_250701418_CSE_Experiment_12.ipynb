```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Generate random data
array = np.random.randint(1, 100, 16)

print("Original Array:")
print(array)

print("Mean:", array.mean())
print("25th Percentile:", np.percentile(array, 25))
print("50th Percentile:", np.percentile(array, 50))
print("75th Percentile:", np.percentile(array, 75))
print("100th Percentile:", np.percentile(array, 100))

# Detect outliers using IQR
def outDetection(array):
    Q1, Q3 = np.percentile(array, [25, 75])
    IQR = Q3 - Q1
    lr = Q1 - 1.5 * IQR
    ur = Q3 + 1.5 * IQR
    return lr, ur

lr, ur = outDetection(array)

print("Lower Range:", lr)
print("Upper Range:", ur)

# Display original distribution
sns.histplot(array, kde=True)
plt.title("Original Data Distribution")
plt.show()

# Remove outliers
new_array = array[(array >= lr) & (array <= ur)]

print("Array After Removing Outliers:")
print(new_array)

sns.histplot(new_array, kde=True)
plt.title("Distribution After Removing Outliers")
plt.show()

# Calculate new ranges
lr1, ur1 = outDetection(new_array)

print("New Lower Range:", lr1)
print("New Upper Range:", ur1)

# Final array
final_array = new_array[
    (new_array >= lr1) & (new_array <= ur1)
]

print("Final Array:")
print(final_array)

sns.histplot(final_array, kde=True)
plt.title("Final Data Distribution")
plt.show()

print("Experiment No. 12 Completed Successfully!")
```
